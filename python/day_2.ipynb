# Python Functions — AI/ML Focus

If your goal is to become strong at **AI/ML and eventually build AI systems**, functions are one of the most important Python concepts to master.

The key is not memorizing syntax. You want to develop the **function mindset**.

---

# 1. The Function Mindset

Think of a function as a **machine**:

```text
Input → Function → Output
```

For example:

```python
def square(x):
    return x * x
```

You give it:

```text
5
```

It produces:

```text
25
```

So whenever you see a task that can be described as:

> "Given X, perform this operation and produce Y."

you should start thinking:

> **"Should this be a function?"**

---

# 2. When SHOULD I use a function?

### Rule 1 — When you repeat logic

Suppose you do this multiple times:

```python
name = "riyan"
print(name.upper())

name = "ali"
print(name.upper())

name = "ahmed"
print(name.upper())
```

You could create reusable logic:

```python
def make_upper(name):
    return name.upper()
```

Then:

```python
print(make_upper("riyan"))
print(make_upper("ali"))
print(make_upper("ahmed"))
```

### Mindset:

> **If I'm writing the same logic more than once, consider a function.**

---

# 3. When you DON'T need a function

Don't turn every two lines of code into a function.

For example:

```python
x = 10
y = 20
print(x + y)
```

There's no strong reason to make:

```python
def add(x, y):
    return x + y
```

unless you're going to reuse that operation or it makes your program clearer.

### Bad mindset

> "Everything must be a function."

### Better mindset

> **"Use a function when giving a name to a reusable piece of logic improves my code."**

---

# 4. Functions are especially important in AI/ML

Imagine you're building a model pipeline:

```text
Raw Data
   ↓
Clean Data
   ↓
Transform Data
   ↓
Train Model
   ↓
Evaluate Model
   ↓
Predict
```

You could represent these as functions:

```python
clean_data()
transform_data()
train_model()
evaluate_model()
predict()
```

This makes your AI project much easier to understand and maintain.

This distinction is important.

```python

```

`name` is a **parameter**.

When you call:

```text
Parameter = placeholder
Argument  = actual value
```

# 7. `return` — Probably the MOST important concept ⭐⭐⭐⭐⭐

Consider:

```python
def add(a, b):
    return a + b
```

When you do:

```python
result = add(10, 20)
```

the function sends `30` back.

```text
10, 20
  ↓
 add()
  ↓
 30
```

You can then use the result:

```python
result = add(10, 20)

print(result)
print(result * 2)
```

Output:

```text
30
60
```

### Important distinction

`print()` displays something.

`return` gives something back to the caller.

For AI/ML code, **`return` is extremely important** because functions need to pass processed data/results to other parts of your pipeline.

---

# 8. `print()` vs `return`

Bad for reusable logic:

```python
def add(a, b):
    print(a + b)
```

Now:

```python
result = add(10, 20)

print(result)
```

gives:

```text
30
None
```

Why?

Because the function printed `30`, but didn't return anything.

Better:

```python
def add(a, b):
    return a + b
```

Now:

```python
result = add(10, 20)
```

`result` contains `30`.

### Mental rule

> **Use `print()` when you want to show something. Use `return` when you want to give something to another part of the program.**

This:


# 16. Scope — ⭐⭐⭐⭐⭐

This is very important.

```python
x = 10

def test():
    x = 20
    print(x)

test()

print(x)
```

Output:

```text
20
10
```

The `x` inside the function is a **local variable**.

The `x` outside is a **global variable**.

Think:

```text
Global scope
    |
    └── Function scope
           |
           └── local variables
```

---

# 17. Avoid unnecessary global variables

Instead of:

```python
data = [10, 20, 30]

def calculate():
    return sum(data)
```

prefer:

```python
def calculate(data):
    return sum(data)
```

Then:

```python
data = [10, 20, 30]

result = calculate(data)
```

Why?

Because the function becomes **independent and reusable**.

This principle is extremely valuable in ML projects.

---

# 18. Functions should generally do ONE job ⭐⭐⭐⭐⭐

Bad:

```python
def process_everything(data):
    # clean data
    # train model
    # evaluate model
    # save model
    # send email
```

Better:

```python
def clean_data(data):
    ...

def train_model(data):
    ...

def evaluate_model(model, data):
    ...

def save_model(model):
    ...
```

Then combine them:

```python
data = clean_data(data)
model = train_model(data)
score = evaluate_model(model, data)
save_model(model)
```

This is called **separation of concerns**.

It becomes extremely important as your AI projects become larger.

---

# 19. Functions can call other functions ⭐⭐⭐⭐⭐

This is how complex systems are built.

```python
def clean_text(text):
    return text.lower().strip()

def tokenize(text):
    return text.split()

def preprocess(text):
    text = clean_text(text)
    words = tokenize(text)

    return words
```

Then:

```python
text = "   MACHINE LEARNING   "

result = preprocess(text)

print(result)
```

Output:

```python
['machine', 'learning']
```

Notice what happened:

```text
preprocess()
     ↓
clean_text()
     ↓
tokenize()
     ↓
result
```

This pattern is **very important for AI systems**.

# Python Lists — The 99% Toolkit

For your **AI/ML + data work**, you do **not** need to memorize every list method. You should master a small set that you'll use constantly.

Think of a list as:

> **An ordered, changeable collection of values.**

```python
numbers = [10, 20, 30, 40]
```

---

## 1. Indexing ⭐⭐⭐⭐⭐

Access one item by position.

```python
numbers = [10, 20, 30, 40]

print(numbers[0])
# 10

print(numbers[2])
# 30
```

Negative indexing:

```python
print(numbers[-1])
# 40
```

Mental model:

```text
[10, 20, 30, 40]
 ↑   ↑   ↑   ↑
 0   1   2   3
```

This is fundamental for algorithms and data processing.

---

# 2. Slicing ⭐⭐⭐⭐⭐

Get part of a list.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
```

Output:

```python
[20, 30, 40]
```

Useful patterns:

```python
numbers[:3]      # first 3
numbers[2:]      # from index 2 onward
numbers[:]       # entire list
numbers[::2]     # every second item
numbers[::-1]    # reverse
```

### AI/ML example

Suppose you have 100 samples:

```python
data = list(range(100))

train = data[:80]
test = data[80:]
```

This is the basic idea behind splitting datasets.

# 4. `.append()` ⭐⭐⭐⭐⭐

# 16. List Comprehension ⭐⭐⭐⭐⭐

This isn't a method, but for your AI/ML goal, **you absolutely need to master it**.

It allows you to create a list from another iterable concisely.

Instead of:

```python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
```

You can write:

```python
squares = [number ** 2 for number in numbers]

print(squares)
```

Output:

```python
[1, 4, 9, 16, 25]
```

### Filtering

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [x for x in numbers if x % 2 == 0]

print(even_numbers)
```

Output:

```python
[2, 4, 6]
```

This is **very important for data cleaning**.

# 18. `zip()` ⭐⭐⭐⭐⭐ for AI/Data

# 19. `enumerate()` ⭐⭐⭐⭐⭐

Use it when you need both the **index and value**.


```python
for i, name in enumerate(names):
```

Output:

```text

```

This is cleaner and extremely common in Python.

# 20. `copy()` — ⭐⭐⭐⭐

Important because lists are **mutable**.

This:

```python
a = [1, 2, 3]
b = a

b.append(4)

print(a)
```

gives:

```python
[1, 2, 3, 4]
```

Why?

Both variables refer to the **same list**.

If you want an independent copy:

```python
a = [1, 2, 3]

b = a.copy()

b.append(4)

print(a)
# [1, 2, 3]

print(b)
# [1, 2, 3, 4]
```

This concept becomes very important when manipulating datasets.

## The most important distinction to memorize

There are four things beginners often confuse:

```python
numbers = [10, 20, 30, 40]
```

**Find a value:**

```python
20 in numbers
```

**Find its position:**

```python
numbers.index(20)
```

**Remove the value:**

```python
numbers.remove(20)
```

**Remove using its position:**

```python
del numbers[1]
```

And for AI/data work, the next major step is learning to combine these with **loops, functions, dictionaries, and list comprehensions**. That is where lists become genuinely powerful for preprocessing and ML data pipelines.

````python
# finding the largest number in the list
def largest_num(numbers):
    if len(numbers) == 0:
        return None
    
    max = numbers[0]
    for number in numbers[1:]:
        if number > max:
            max = number
    return max

numbers = [-5, -2, -9] 
largest = largest_num(numbers)
if largest:
    print(largest)
else:
    print(f"list is empty")
````

````python
# removing the duplicates and preserving the order
# test = [1, 2, 2, 3, 1, 4]
test = []

def remove_duplicates(numbers):
    
    # edge case if list is empty
    if len(numbers) == 0:
        return None
    
    clean_list = []
    for num in numbers:
        if not (num in clean_list):
            clean_list.append(num)
    return clean_list

result = remove_duplicates(test)

print(test)
if result:
    print(result)
else:
    print("list is empty")
````

````python
#  checking palindrome for a string

# text_1 = "Never odd or even"
text_1 = ""

def is_palindrome(text):
    
    # for edge cases
    if not text:
        return False
    
    # removing spaces and cases
    text = text.lower().replace(" ","")
    
    # start and ending index
    start = 0
    end = len(text) - 1
    
    # looping until the end
    while start < end:
        if not (text[start] == text[end]):
            return False
        
        # changing indexes
        start += 1
        end -= 1
        
    # loop ends it means the 
    return True

test = is_palindrome(text_1)
print(test)
````

````python
# word frequency 
sentence = "The cat and the hat"

def frequency_counter(sentence):
    # cleaning the sentence
    text = sentence.lower().strip().replace("  "," ")
    words = text.split()
    
    # making dictionary
    frequency = {}
    for word in words:
        if word not in frequency.keys():
            frequency[word] = 1
        else:
            frequency[word] += 1
    return frequency

result = frequency_counter(sentence)
print(result)
    
    
````

````python
numbers = [ 1,2,3,4,5]

def second_largest(values):
    
    if len(values) == 0:
        return None
    
    numbers = values.copy()
    
    distinct_values = set()
    max = 0
    while numbers:
        max = numbers[0]
        for number in numbers:
            if number > max:
                max = number
        # found the maximum value
        distinct_values.add(max)
        numbers.remove(max)
        
    if len(distinct_values) < 2:
        return None
    else:
        sorted_distinct = sorted(distinct_values)
        return sorted_distinct[-2]
    

value = second_largest(numbers)
print(value)
````
