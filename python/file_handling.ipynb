```python
import json
from pathlib import Path 

def load_data(path):
    try:
        with open(path) as f:
            return json.load(f)
    except FileNotFoundError:
        return []
    except json.JSONDecodeError:
        return []
    
def save_data(path,records):
    with open(path, "w",encoding="utf-8") as f:
        json.dump(records, f,indent = 2)

def append_data(path,data):
    notes = load_data(path)
    notes.append(data)
    save_data(path, notes)
    
def read_data(path):
    data = load_data(path)
    if not data:
        return "data not found"
    print(data)
    
        
```

```python
path = Path("data") / "students.json"
path.parent.mkdir(exist_ok= True)

while(True):
    print("1) Input data")
    print("2) show data")
    print("3) exit")

    choice = input("Enter your choice:")

    if choice == "1":
        name = input("enter your name:") 
        goal = input("enter your goal:") 
        append_data(path, {"name":name ,"goal":goal} )
    elif choice == "2":
        read_data(path)
    else:
        break
```

```python
# 1,100

def ask_in(prompt, low , high):
    """this functions asks for int value between range and handles errors"""

    try:
        value = int(prompt)
    except ValueError as e:
        print(f"Error: Please enter a valid value")
        return None
    
    if value <= high and value >= low:
        return value
    else:
        return None
    
        
    
while True:
    prompt = input("Enter a value:")
    response = ask_in(prompt , 1, 100)
    if response:
        print(f"value is : {response}")
        break
    else:
        print("try again:")
    
    
```

```python
user = {"profile": {"role": "student"}}

def value(data, *keys):
    key, value = keys
    try:
        return data[key][value]
    except KeyError:
        return None
        
print(value(user, "profile","role"))
print(value(user, "profile","student"))
```

```python
path = "data/notes.txt"
def numbers(path):
    file_data = ""
    numbers = []
    try:
        with open(path, "r") as f:
            file_data = f.read()
    except FileNotFoundError:
        print(f"file is not found!")
        return []

    # cleaning the file
    file_data = file_data.split("/")
    for item in file_data:
        item = item.strip()
        # if item == "" or item == "":
        #     continue
        try:
            item = float(item)
            numbers.append(item)
        except ValueError:
            continue
            
    return numbers

lis = numbers(path)
print(lis)
```
