```python
words = ["banana", "kiwi", "apple"]
print(sorted(words))                       # alphabetical
print(sorted(words, key=len))              # by length: kiwi, apple, banana
print(sorted(words, key=len, reverse=True))

students = [
    {"name": "Ali", "dept": "CS", "marks": 80},
    {"name": "Sara", "dept": "AI", "marks": 92},
    {"name": "Omar", "dept": "DI", "marks": 85},
]

# by marks, highest first
print(sorted(students, key=lambda s: s["marks"], reverse=True))

# two levels: dept A-Z, then marks highest first (minus flips numbers)
print(sorted(students, key=lambda s: (s["dept"], -s["marks"])))

# the same idea with max and min
print(max(students, key=lambda s: s["marks"])["name"])    # Sara
```

```python
value = "hldsfkjas"
print(value[0::3])
```

```python
def make_chunks(value, size):
    str_len = len(value)
    lis = []
    for i in range(0,str_len,size):
        lis.append(value[i:i + size])
        
    return lis

result = make_chunks(value,3)
print(result)
```
