```python
words = ["banana", "kiwi", "apple"]
print(sorted(words))                       # alphabetical
print(sorted(words, key=len))              # by length: kiwi, apple, banana
print(sorted(words, key=len, reverse=True))

students = [
    {"name": "Ali", "dept": "CS", "marks": 80},
    {"name": "Sara", "dept": "AI", "marks": 92},
    {"name": "Omar", "dept": "DI", "marks": 85},
]

# by marks, highest first
print(sorted(students, key=lambda s: s["marks"], reverse=True))

# two levels: dept A-Z, then marks highest first (minus flips numbers)
print(sorted(students, key=lambda s: (s["dept"], -s["marks"])))

# the same idea with max and min
print(max(students, key=lambda s: s["marks"])["name"])    # Sara
```

```python
def make_chunks(value, size):
    str_len = len(value)
    lis = []
    for i in range(0,str_len,size):
        lis.append(value[i:i + size])
        
    return lis

result = make_chunks(value,3)
print(result)
```

### Problem 02


```python
def trim_ends(lis):
    if len(lis) < 3:
        return []
    else:
        return lis[1:-1]
```

```python
print(trim_ends([1,2,3,4]))
print(trim_ends([3,6,4]))
```

```python
assert trim_ends([1, 2, 3, 4]) == [2, 3]     # normal
assert trim_ends([]) == []                   # empty
assert trim_ends([1, 2]) == []   
```

### Problem 03


```python
# a function that will clean a sentence, break it into list of words,
# lowercase the words of atleast three letters

def clean_sentence(sentence):
    collection = [word.lower().strip(".,!") if len(word) >= 3 else word.strip(".,!") for word in sentence.split()]
    return collection

```

```python
clean_sentence("The cat is in the Hat!")
```

```python
def top_n(dictionary,top):
    result = sorted(dictionary.items(), key= lambda item:(-item[1],item[0]))
    return result[:top]
```

```python
top_n({"Ali": 80, "Sara": 92, "Omar": 80}, 2)
```
