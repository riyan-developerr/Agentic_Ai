```python
# problem number one

number = int(input("enter a number:"))
for val in range(1,number + 1):
    if val % 2 == 0:
        print(f"{val} even")
    else:
        print(f"{val} odd")
```

```python
# Sum of digits

number = int(input("enter a number"))
copy = number
sum = 0
while(number > 0):
    rem = number % 10
    sum += rem
    number = number // 10
    
print(f"the sum of {copy} is: {sum}")
```

```python
# fizz buzz thing
for num in range(1,31):
    if num % 3 == 0 and num % 5 == 0:
        print("FizzBzuz")
    elif num % 3 == 0:
        print("Fizz")
    elif num % 5 == 0:
        print("Buzz")
    else:
        print(num)
```

```python
# counting vowels in a sentence
sentence = input("enter a sentence:")
counter = 0
for ch in sentence:
    if ch.lower() == "a" or ch.lower() == "e" or ch.lower() == "i" or ch.lower() == "o" or ch.lower() == "u":
        counter += 1
        
print(f"total vowels in the sentence: {counter}")
```

```python
# password system with three attempts
password = "python123"
granted = False
attempts = 0

while(attempts < 3):
    
    user_try = input("enter password:")
    attempts += 1
    
    if user_try == password:
        print("access granted")
        granted = True
        break
        
    elif attempts < 3:
        print(f"try again remaining attempts: {3 - attempts}")
           
    
if not granted :
    print("Account locked")
```

```python
# star pattern right angled triangle
n = int(input("enter the number:"))
for i in range(n):
    for j in range(0, i + 1):
        print(" * ",end = "")
    print()
        
```
