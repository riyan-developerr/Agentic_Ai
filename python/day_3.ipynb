```python
# merging two dictionaries

basic = {
    "name": "Ali",
    "company": "ABC"
}

classification = {
    "industry": "Software",
    "priority": "high"
}

# both approaches works fine
# merged = {**basic,**classification}
merged = basic | classification

print(len(merged))
print(merged)
```

```python
students = [
    {"name": "Ali", "department": "AI"},
    {"name": "Ahmed", "department": "CS"},
    {"name": "Riyan", "department": "AI"},
    {"name": "Usman", "department": "CS"}
]

faculty = {}
for student in students:
    if student["department"] not in faculty:
        faculty[student["department"]] = [student["name"]]
    else:
        faculty[student["department"]].append(student["name"])

print(faculty)
        
    
```

```python
collection = ["hi", "cat", "dog", "to"]
dictionary = {}

for item in collection:
    if len(item) not in dictionary:
        dictionary[len(item)] = [item]
    else:
        dictionary[len(item)].append(item)
        
print(dictionary)
        
    
```

```python
sentence_1 = "the cat sat"
sentence_2 = "the dog sat"

sentence_1 = sentence_1.lower().split()
sentence_2 = sentence_2.lower().split()


set_1 = set(sentence_1)
print(set_1)
set_2 = set(sentence_2)
print(set_2)

both = set_1 | set_2
print(both)
first_unique = set_1 - set_2
print(first_unique)
```

```python
data = {"a": 1, "b": 2, "c": 1, "d":1, "f":2}
new_data = {}
for key, value in data.items():
    if value not in new_data:
        new_data[value] = [key]
    else:
        new_data[value].append(key)

print(new_data)
        
```

```python
collection = ["red apple", "green apple", "red car"]
mapping = {}

for index,sen in enumerate(collection):
    sen = sen.lower().split()
    for word in sen:
        if word not in mapping:
            mapping[word] = [index]
        else:
            mapping[word].append(index)
            
print(mapping)
            
```
